```python
%%writefile fizzbuzz.py

import numpy as np

numbers = np.arange(1,101)

for number in numbers:
    if number % 3 == 0 and number % 5 == 0:
        output = "FizzBuzz"
    elif number % 3 == 0:
        output = "Fizz"
    elif number % 5 == 0:
        output = "Buzz"
    else:
        output = number
    print(f"{number}:{output}")
```

```python
%run fizzbuzz.py
```

### Extensions

```python
%%writefile fizzbuzz_extension_A.py

import numpy as np

numbers = np.arange(1, 101)

for number in numbers:
    output = ""

    if number % 3 == 0:
        output += "Fizz"
    if number % 5 == 0:
        output += "Buzz"
    if number % 7 == 0:
        output += "Fang"
    if number % 11 == 0:
        output += "Bang"

    if output == "":
        output = number

    print(f"{number}:{output}")
```

```python
%run fizzbuzz_extension_A.py
```

```python
%%writefile fizzbuzz_extension_B.py

import numpy as np

def fizzbuzz(limit, rules):
    numbers = np.arange(1, limit + 1)

    for number in numbers:
        output = ""
        for factor in rules:
            if number % factor == 0:
                output += rules[factor]

        if output == "":
            output = number

        print(f"{number}:{output}")

if __name__ == "__main__":
    rules = {3: "Fizz", 5: "Buzz", 7: "Fang", 11: "Bang"}
    fizzbuzz(100, rules)
```

```python
%run fizzbuzz_extension_B.py
```

```python
%%writefile fizzbuzz_extension_C.py
import argparse
import numpy as np

def fizzbuzz(limit, rules):
    numbers = np.arange(1, limit + 1)

    for number in numbers:
        output = ""

        for factor in rules:
            if number % factor == 0:
                output += rules[factor]

        if output == "":
            output = number

        print(f"{number}:{output}")

if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--limit", type=int, default=100)
    args = parser.parse_args()

    rules = {3: "Fizz", 5: "Buzz", 7: "Fang", 11: "Bang"}
    fizzbuzz(args.limit, rules)
```

```python
%run fizzbuzz_extension_C.py --limit 300
```
